```python
def f(x):
  return x*x - 3

def df(x):
  return 2*x

x0 = 1.5
error = 0.0000000001

def newton(f, df, x0, error):
  x = x0
  while np.abs(f(x)) > error:
    x = x - f(x)/df(x)
  return x
```

```python
vec = np.array([3, 3])
matrix = np.array([[-1, 0], [0, 1]])
vec_2 = matrix @ vec
print(vec_2)
```

```python

# Creating arrow
x_pos = [0, 0]
y_pos = [0, 0]
x_direct = [5, vec_2[0]]
y_direct = [5, vec_2[1]]

# Creating plot
fig, ax = plt.subplots(figsize = (7, 4))
ax.quiver(x_pos, y_pos, x_direct, y_direct, scale=5)
ax.set_title('Quiver plot with one arrow')

# Show plot
plt.show()
```
